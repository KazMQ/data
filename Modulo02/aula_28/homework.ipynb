ATIVIDADE 

O Secretário de Segurança Pública solicitou uma análise a fim de saber, se há relação entre lesões dolosas e lesões com morte. Se isto pode ser observado com base nos dados históricos. Se sim, ele pede p/ estimar, de alguma forma, quantas lesões com morte podem ser previstas, quando as lesões dolosas forem em torno de 40.000, 50.000 e 60.000 em 3 meses.  

Isto norteará uma nova estratégia de segurança para o estado. 

```python
import pandas as pd
import numpy as np

data_path = 'https://www.ispdados.rj.gov.br/Arquivos/BaseDPEvolucaoMensalCisp.csv'

#Obtendo dados
try:
    print("Downloading data...")
    df_ocorrencias = pd.read_csv(data_path, sep=';', encoding='iso-8859-1', low_memory=False)

    print(df_ocorrencias.head())
except Exception as e:
    print(f"An error occurred: {e}")
```

```python
#Delimitando as Variáveis e totalizando
try:
    df_lesao = df_ocorrencias [['cisp', 'lesao_corp_dolosa','lesao_corp_morte']]

    #totalizando os dados
    df_total_lesaos = df_lesao.groupby('cisp').sum(['lesao_corp_dolosa', 'lesao_corp_morte']).reset_index()
    #print(df_ocorrencias.head())

except Exception as e:
    print(f"An error occurred while processing data: {e}")
    exit()
                        
```

```python
#Calculando Correlação

try:
    df_total_lesao_cut = (
        df_total_lesaos[df_total_lesaos['lesao_corp_dolosa'] > np.percentile(df_lesao['lesao_corp_morte'], 95)] 
    )

    df_total_lesao_cut = (
        df_total_lesaos
            [df_total_lesaos['lesao_corp_morte'] > np.percentile(df_lesao['lesao_corp_morte'], 99)]
    )

    #Arrays
    array_lesao = np.array(df_total_lesao_cut['lesao_corp_dolosa'])
    array_mort = np.array(df_total_lesao_cut['lesao_corp_morte'])
    #Calculando Correlação
    correlation = np.corrcoef(array_lesao, array_mort)[0, 1]
except Exception as e:
    print(f"An error occurred while filtering data: {e}")
    exit()
```

```python
#Aplicando o modelo de regressão linear
try:
    print('Iniciando a regressão linear...')
    from sklearn.model_selection import train_test_split
    X_train, X_test, y_train, y_test = train_test_split(
        array_lesao, #Variável independente
        array_mort, #Variável dependente
        test_size=0.2, #80% treino, 20% teste
        random_state=42 #Para reprodutibilidade
    )

    from sklearn.preprocessing import StandardScaler
    scaler = StandardScaler()

    X_train = scaler.fit_transform(X_train.reshape(-1, 1))
    X_test = scaler.transform(X_test.reshape(-1, 1))

    from sklearn.linear_model import LinearRegression
    modelo = LinearRegression()
    modelo.fit(X_train, y_train) #treinamento do modelo

    r2_score = modelo.score(X_test, y_test) #R² do modelo
    print(f"R² do modelo: {r2_score:.2f}")

except Exception as e:
    print(f"An error occurred during model training: {e}")
    exit()
```

```python
#Previsão
try:
    #Array de previsões 35000, 43000 e 58000
    array_lesao_predict = np.array([40000, 50000, 60000])

    #Normalizando os dados de entrada
    array_lesao_predict_scaled = scaler.transform(array_lesao_predict.reshape(-1, 1))

    #Fazendo Previsões
    mort_pred = modelo.predict(array_lesao_predict_scaled)
    print("Previsões de recuperação de veículos: ", mort_pred)

    

except Exception as e:
    print(f"An error occurred while preparing prediction data: {e}")
    exit()
```

```python
##################

# AVALIANDO O MODELO

# P/ Rodar a Tkinter p/ plotar o painel como no Jupyter
# import matplotlib
# matplotlib.use('TkAgg')
import matplotlib.pyplot as plt
import seaborn as sns

try:
    print('Avaliando o modelo de previsões...')

    plt.subplots(2, 2, figsize=(15, 5))
    plt.suptitle('Avaliação do modelo de regressão')


    #### POSIÇÃO 1: Gráfico de Dispersão entre os arrays
    # pip install seaborn
    plt.subplot(2, 2, 1)

    sns.regplot(x=array_lesao, y=array_mort)
    plt.title('Gráfico de dispersão')
    plt.xlabel('Lesão corporal dolosa')
    plt.ylabel('Lesão corporal com morte')

    # Posicionar o texto no início do gráfico
    plt.text(min(array_lesao),
             max(array_mort) * 0.9,  # Ajuste a posição vertical conforme necessário
             f'Correlação: {correlation}',
             fontsize=10)


    #### POSIÇÃO 2: Gráfico de dispersão entre os dados reais e previsto
    plt.subplot(2, 2, 2)

    # Testar o modelo preditivo nos dados de X_teste
    # Saída na escal original dos dados
    y_pred = modelo.predict(X_test)

    # Retornar os dados de teste para escala real
    X_test = scaler.inverse_transform(X_test)

    # Gráfico de dispersão sem a linha de regressão
    plt.scatter(X_test, y_test, color='blue', label='Dados reais')
    plt.scatter(X_test, y_pred, color = 'red', label='Previsões')

    plt.title('Dados reais x previstos')
    plt.xlabel('Lesão corporal dolosa (testes)')
    plt.ylabel('Lesão corporal com morte (testes)')

    plt.legend()


    ##### POSIÇÃO 3: Resíduos
    plt.subplot(2, 2, 3)

    # RESÍDUOS: Diferença entre valores reais e previstos (y_test - y_pred)
    # Análise dos resíduos ajuda a verificar, se o modelo está capturando bem
    # os padrões dos dados. 
    # Bons modelos têm resíduos próximos de 0.
    # Os dados devem ser distribuídos aleatoriamente.
    # Se tiverem um padrão, o modelo não é confiável
    residuos = y_test - y_pred

    # plotar em gráfico de dispersão
    plt.scatter(y_pred, residuos)

    # adicioanar uma linha constante no 0
    plt.axhline(y=0, color='black', linewidth=2)

    plt.title('Resíduos')
    plt.xlabel('Previsões')
    plt.ylabel('Resíduos')


    #### POSIÇÃO 4: Dispersão dos valores simulados
    plt.subplot(2, 2, 4)
    
    plt.scatter(array_lesao_predict, mort_pred)

    plt.title('Previsão de lesões simuladas')
    plt.xlabel('Lesão corporal dolosa simulada')
    plt.ylabel('Lesão corporal com morte prevista')

    plt.tight_layout()
    plt.show()

except ImportError as e:
    print("Erro ao avaliar o modelo: ", e)
    exit()
```
