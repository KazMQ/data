ATIVIDADE 

O Secretário de Segurança Pública solicitou uma análise a fim de saber, se há relação entre lesões dolosas e lesões com morte. Se isto pode ser observado com base nos dados históricos. Se sim, ele pede p/ estimar, de alguma forma, quantas lesões com morte podem ser previstas, quando as lesões dolosas forem em torno de 40.000, 50.000 e 60.000 em 3 meses.  

Isto norteará uma nova estratégia de segurança para o estado.
